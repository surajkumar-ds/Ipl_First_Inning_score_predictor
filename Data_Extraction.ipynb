```python
import pandas as pd
import numpy as np 
from yaml import safe_load
import os
from tqdm import tqdm
```

```python
filenames = []
for file in os.listdir('data'):
    filenames.append(os.path.join('data',file))
```

```python
filenames[0:5]
```

```python
final_df = pd.DataFrame()
counter = 1

for file in tqdm(filenames):
    with open(file, 'r') as f:
        df = pd.json_normalize(safe_load(f))
        df['match_id'] = counter

        final_df = pd.concat([final_df, df], ignore_index=True)
        counter += 1

final_df
```

```python
count = 1
delivery_df = pd.DataFrame()

for index, row in matches.iterrows():

    if count in [75,108,150,180,268,360,443,458,584,748,982,1052,1111,1226,1345]:
        count += 1
        continue

    count += 1

    # Handle missing or invalid innings data
    innings = row['innings']

    if not isinstance(innings, list) or len(innings) == 0:
        continue

    first_innings = innings[0]

    if not isinstance(first_innings, dict):
        continue

    if '1st innings' not in first_innings:
        continue

    deliveries = first_innings['1st innings'].get('deliveries', [])

    if not isinstance(deliveries, list):
        continue

    # Keep your existing list definitions here
    ball_of_match = []
    batsman = []
    bowler = []
    runs = []
    player_of_dismissed = []
    teams = []
    batting_team = []
    match_id = []
    city = []
    venue = []
    is_legal_ball = []

    for ball in deliveries:
        # Keep your existing inner-loop code here
        for key in ball.keys():
            match_id.append(count)
            batting_team.append(row['innings'][0]['1st innings']['team'])
            teams.append(row['info.teams'])
            ball_of_match.append(key)
            batsman.append(ball[key]['batsman'])
            bowler.append(ball[key]['bowler'])
            runs.append(ball[key]['runs']['total'])
            extras = ball[key].get('extras', {})

            is_legal_ball.append(
        0 if ('wides' in extras or 'noballs' in extras) else 1
        )
            city.append(row['info.city'])
            venue.append(row['info.venue'])

            try:
                player_of_dismissed.append(
                    ball[key]['wicket']['player_out']
                )
            except (KeyError, TypeError):
                player_of_dismissed.append('0')

    loop_df = pd.DataFrame({
    'match_id': match_id,
    'teams': teams,
    'batting_team': batting_team,
    'ball': ball_of_match,
    'runs': runs,
    'player_dismissed': player_of_dismissed,
    'city': city,
    'venue': venue,
    'is_legal_ball': is_legal_ball
})

    delivery_df = pd.concat(
        [delivery_df, loop_df],
        ignore_index=True
    )

delivery_df
```

```python
delivery_df['teams'] = [
    row['info.teams'].iloc[0]
    for row in delivery_df['teams']
]
```

```python
delivery_df
```

```python
# getting the bowling team name
# Getting the bowling team name
def bowl(row):
    for team in row['teams']:
        if team != row['batting_team']:
            return team

delivery_df['bowling_team'] = delivery_df.apply(bowl, axis=1)

delivery_df.head()
```

```python
# removing the team column
delivery_df.drop(columns=['teams'],inplace=True)
```
